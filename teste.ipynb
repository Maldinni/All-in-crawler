```python
import pandas as pd
from pathlib import Path

# 1) aponte para um shard (ou pegue o mais recente automaticamente)
SHARDS_DIR = Path("data/output")
csv_files = sorted(SHARDS_DIR.glob("shard_*.csv"))

if not csv_files:
    raise FileNotFoundError(f"Nenhum shard_*.csv encontrado em {SHARDS_DIR.resolve()}")

shard_path = csv_files[-1]  # mais recente
print("Lendo:", shard_path)

# 2) carrega o CSV
df = pd.read_csv(shard_path)

# 3) ajustes pra visualizar melhor no notebook
pd.set_option("display.max_colwidth", 120)  # aumenta a largura do texto
pd.set_option("display.max_columns", 50)

# 4) escolha uma “tabela amigável” (com preview do texto)
table = df.copy()
if "text" in table.columns:
    table["text_preview"] = table["text"].fillna("").str.replace(r"\s+", " ", regex=True).str[:300] + "..."
    table = table.drop(columns=["text"])

# 5) reordena colunas (se existirem)
preferred = ["source", "publish_date", "title", "url", "text_preview", "authors", "keywords"]
cols = [c for c in preferred if c in table.columns] + [c for c in table.columns if c not in preferred]
table = table[cols]

# 6) mostra a tabela (primeiras 50 linhas)
table.head(50)
```
